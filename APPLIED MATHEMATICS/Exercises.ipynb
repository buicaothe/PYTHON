```python
# 1. NUMBER SEQUENCE

# a) Arithmetic sequence has a first member a_1 and a common difference d (erotusluku).
# Choose a_1 and d to be the last two non-zero digits from your student number. [a_1 = 7; d = 4]
# Create the general term a_n of this arithmetic sequence. [a_n = a_1+d(n-1) --> a_n = 7+4(n-1) --> a_n = 4n+3]
# Create and print a list of the first n (where n is your age) terms of this sequence with a_n. [n = 45]
# Plot the graph of these first members of the sequence.

# b) Geometric sequence has a first member a_1 and a common ratio q (suhdeluku).
# Choose a_1 to be the first digit of your age.
# Choose q to be the last digit of your student number, which is not zero or one.
# Create the general term a_n of this geometric sequence.
# Solve the index n (järjestysluku) of member 15740.
# The index n might not be an integer, so print it as a floating point number.
# *===================================================================================

# Creating symbolic variable n (which is now an integer)
n = sy.symbols('n', integer=True)

# a) ARITHMETIC SEQUENCE:
# General term a_n with (a_1 = 7; d = 4):
a_n = 4*n+3

# Calculating the member a_45 with subs-method:
a_45 = a_n.subs(n, 45)
print('- Value of (last) member 45: ', a_45)

# Calculating the first 45 members as a Python-list with subs-method
members1_45 = [a_n.subs(n, i) for i in range(1, 46)]

print('- First 45 members (subs-method): ', '\n', members1_45)

# Calculating the first 45 members with lambdifying
# transforming sympy expression a_n into numpy lambda-function f_n
f_n = sy.lambdify(n, a_n, "numpy")
list45 = np.arange(1, 46)  # creating numpy-ndarray of integers 1,2,...,44,45
# lambda-functions can take a numpy-ndarray as an input
members1_45 = f_n(list45)
# the output is also a numpy-ndarray
print('- First 45 members (lambdifying method): ''\n', members1_45)

# Solving equations
an = int(input('Enter the Value of certain a_n = '))
equation = sy.Eq(a_n, an) # creating a sympy-equation "a_n = input" with Eq-function
index = sy.solve(equation, n) # solving the equation "a_n = input" with respect to n
if index:
    print(f'- The index of the member {an} is ',index) # the solution of solve-function will be a list-type object
    print(f'- Checking by replace n = {index[0]} into a_n, we have a_{index[0]} = ',a_n.subs(n,index[0])) # the solution can be verified by substituting the index into a_n
else:
    print(f'- The value {an} is NOT a member of the arithmetic sequence (n is not an integer).')

# Plot the graph of these first 45 members of the sequence:
# drawing a scatter-plot with matplotlib for a_n
plt.scatter(list45, members1_45, color='green')
plt.title("ARITHMETIC SEQUENCE")  # title of the whole plot
plt.xlabel("n")  # the title of x-axis
plt.ylabel("a_n")  # the title of y-axis
plt.grid(True)  # drawing the grid
plt.show()  # showing the actual plot

# b)

```

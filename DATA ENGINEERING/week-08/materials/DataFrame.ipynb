```python
import pandas as pd

namespaces = {
    "ns": "https://example.edu/deai/readings"
}

# Lấy chi tiết từng reading
df = pd.read_xml(
    "readings.xml",
    xpath="//ns:reading",
    namespaces=namespaces
)
df
```

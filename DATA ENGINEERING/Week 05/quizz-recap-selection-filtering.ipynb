# Week 5 — Selection and filtering

The table is the Star Wars characters file (96 rows). You will pick columns and rows in two ways:

- `[]` selects a column, a list of columns, or the rows where a condition is true
- `.loc` selects rows and columns by label in one step

Basic selections come first. The last examples combine a row filter with a column list.


## Load the characters


````python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/planetacomputer/"
    "starwars-full/main/characters.csv"
)
df = pd.read_csv(url)
df.head()
````

````python
df.shape
````

## One column, then several columns

`df["name"]` returns a Series: one column, with the row labels kept.

`df[["name", "species"]]` returns a DataFrame. The inner list is the signal that you want a table, even when the list holds one name: `df[["name"]]`.


````python
df["name"].head() # return a pandas Series, not a DataFrame
````

````python
df[["name"]].head() # return a DataFrame, not a pandas Series
````

````python
df[["name", "species", "homeworld"]].head() # return a DataFrame, not a pandas Series
````

## Filter rows with `[]`

A comparison such as `df["height"] > 2` is a column of `True` and `False`, one value per row. Putting that column inside `[]` keeps the rows where it is `True`.


````python
tall = df["height"] > 2 # a condition: return in Boolean value (True or False) for the condition 'tall', not a number!
tall.head()
````

````python
df[tall][["name", "species", "height"]] # Fillter row with [] only: condition "tall" return in Boolean value (True or False), not a number! 
# If 'tall' is True, the dataframe will show 'name', 'species', 'height'. if False, not showing any row.
# return a DataFrame
````

## The same filter with `.loc`

`.loc` takes the rows and the columns together: `.loc[rows, columns]`.

`tall` still chooses the rows. The list chooses the columns.


````python
df.loc[tall, ["name", "species", "height"]] # condition with loc: [row conditions, 'list' of 'columns']
# return a DataFrame (loc)
type(df.loc[0, ["name", "species", "height"]])
````

One cell can also pick a single label. Row label `0` is Luke Skywalker. A single column name returns one value.


````python
df.loc[0, "name"] # loc[index = 0, 'column name'] # can also be 'iloc'. Return a string, a Scalar value from one cell in Series
# note: 'Integer-based' not same as 'Label-based' index

df.loc[0, ["name"]] # loc[index = 0, 'column name'] # can also be 'iloc'. Return a string, a Scalar value from one cell in Series
# note: 'Integer-based' not same as 'Label-based' index

# check type:
print(type(df.loc[0, "name"]))   # Return a string!
print(type(df.loc[0, ["name",'height']])) # Return a Series!
print(type(df.loc[[0], ["name",'height']])) # Return a DataFrame! when putting the row in [...]
df.loc[[0], ["name",'height']] # Return a DataFrame! when putting the row in [...]. Cannot use 'iloc' because 'columns' are in Label not in Integer-based Index.
````

````python
df.loc[0, ["name", "species", "height", "homeworld"]] # with Label-based index row: return a Series with column names as Label-based index.
a = df.loc[0, ["name", "species", "height", "homeworld"]]

print(type(a))
a['name'] # use Label-based Index to obtain the value in Series.
a.iloc[0] # = use Positional/Integer-based Index to obtain the value.
# a.loc[0] --> 'loc' not working with this Series' Integer-based Index.

b = df.loc[[0], ["name", "species", "height", "homeworld"]] # row in [...] and cloumns in [...] return in DataFrame
print(type(b))
b['name'] # use Label-based Index to obtain the value in DataFrame.
b.iloc[0] # = use Positional/Integer-based Index to obtain the value.
b.loc[0] # = use Positional/Integer-based Index to obtain the value. 
# for DataFrame, both iloc and loc are working with Integer-based Index.
````

## Two conditions

Each comparison sits in its own parentheses. `&` means both are true. `|` means either one is true.

Humans from Tatooine:


````python
from_tatooine = (df["species"] == "Human") & (df["homeworld"] == "Tatooine") # assign conditions to single variable 'from_tatooine'/
df.loc[from_tatooine, ["name", "homeworld", "height"]] # return a new DAtaFrame with conditions, only 3 columns selected include 'name', homeworld, and 'height'
````

## In depth

**A slice of labels includes the stop.** `df.loc[0:2, "name"]` returns labels 0, 1, and 2: Luke, Leia, and Vader.

**A slice with `[]` uses positions and leaves the stop out.** `df[0:2]` returns the first two rows.


````python
df.loc[0:2, "name"] # loc: [rows: use Positional/Integer Index, 'column name']
print(df.loc[0:2, "name"])
type(df.loc[0:2, "name"]) # 'pandas Series'
type(df.loc[0:2, ["name"]]) # 'DataFrame'
df.loc[0:2, ["name"]]
````

````python
df[0:2]["name"] # use only [..., ...]: [rows: use Positional/Integer Index, 'column name']. Result in a Series
df[0:2][["name"]] # use only [..., ...]: [rows: use Positional/Integer Index, 'column name'in [...]]. Result in a DataFrame (columns in [])
````

A list of labels picks those rows and skips the ones in between.


````python
df.loc[[0, 2, 5], ["name", "species", "height"]] # list of row indices [0,2,5]
````

`.isin` keeps rows whose value is in a list. Here, red eyes or yellow eyes, with the name beside the color.


````python
red_or_yellow = df["eye_color"].isin(["Red", "Yellow"]) # conditions: eye_color = 'Red' or 'Yellow'.
red_or_yellow = (df['eye_color']=='Red') | (df['eye_color']=='Yellow') # conditions: eye_color = 'Red' or 'Yellow'.
df.loc[red_or_yellow, ["name", "species", "eye_color"]]
df[red_or_yellow][["name", "species", "eye_color"]]
````

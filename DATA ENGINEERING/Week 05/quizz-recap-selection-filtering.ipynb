# Week 5 — Selection and filtering

The table is the Star Wars characters file (96 rows). You will pick columns and rows in two ways:

- `[]` selects a column, a list of columns, or the rows where a condition is true
- `.loc` selects rows and columns by label in one step

Basic selections come first. The last examples combine a row filter with a column list.


## Load the characters


````python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/planetacomputer/"
    "starwars-full/main/characters.csv"
)
df = pd.read_csv(url)
df.head()
````

````python
df.shape
````

## One column, then several columns

`df["name"]` returns a Series: one column, with the row labels kept.

`df[["name", "species"]]` returns a DataFrame. The inner list is the signal that you want a table, even when the list holds one name: `df[["name"]]`.


````python
df["name"].head()
````

````python
df[["name", "species", "homeworld"]].head()
````

## Filter rows with `[]`

A comparison such as `df["height"] > 2` is a column of `True` and `False`, one value per row. Putting that column inside `[]` keeps the rows where it is `True`.


````python
tall = df["height"] > 2
tall.head()
````

````python
df[tall][["name", "species", "height"]]
````

## The same filter with `.loc`

`.loc` takes the rows and the columns together: `.loc[rows, columns]`.

`tall` still chooses the rows. The list chooses the columns.


````python
df.loc[tall, ["name", "species", "height"]]
````

One cell can also pick a single label. Row label `0` is Luke Skywalker. A single column name returns one value.


````python
df.loc[0, "name"]
````

````python
df.loc[0, ["name", "species", "height", "homeworld"]]
````

## Two conditions

Each comparison sits in its own parentheses. `&` means both are true. `|` means either one is true.

Humans from Tatooine:


````python
from_tatooine = (df["species"] == "Human") & (df["homeworld"] == "Tatooine")
df.loc[from_tatooine, ["name", "homeworld", "height"]]
````

## In depth

**A slice of labels includes the stop.** `df.loc[0:2, "name"]` returns labels 0, 1, and 2: Luke, Leia, and Vader.

**A slice with `[]` uses positions and leaves the stop out.** `df[0:2]` returns the first two rows.


````python
df.loc[0:2, "name"]
````

````python
df[0:2]["name"]
````

A list of labels picks those rows and skips the ones in between.


````python
df.loc[[0, 2, 5], ["name", "species", "height"]]
````

`.isin` keeps rows whose value is in a list. Here, red eyes or yellow eyes, with the name beside the color.


````python
red_or_yellow = df["eye_color"].isin(["Red", "Yellow"])
df.loc[red_or_yellow, ["name", "species", "eye_color"]]
````

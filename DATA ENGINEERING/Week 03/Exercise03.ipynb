```python
import pandas as pd
```

```python
df1 = pd.read_csv("Afghanistan.csv")
df2 = pd.read_csv("Albania.csv")
df3 = pd.read_csv("Algeria.csv")
df4 = pd.read_csv("Angola.csv")
df5 = pd.read_csv("Armenia.csv")
```

```python
# --------------------------------------------------------------
# A. Use pandas.concat to combine 4 countries from the Maddisson CSV files in one df.
# --------------------------------------------------------------

df = pd.concat([df1,df2,df3,df4])
df
```

```python
# --------------------------------------------------------------
# B. Find statistical data for four different countries and use `pandas.concat()` to combine the results into a single DataFrame.
# --------------------------------------------------------------
def statdata(dfname,country):
    minyear = dfname["Year"].min()
    maxyear = dfname['Year'].max()
    gdpminyear = round(dfname.loc[dfname["Year"].idxmin(),'GDP']/1e9, 2)
    gdpmaxyear = round(dfname.loc[dfname["Year"].idxmax(),'GDP']/1e9, 2)
    avegdp = round(dfname['GDP'].mean()/1e9,2)
    n = maxyear - minyear
    cagr = round(((gdpmaxyear/gdpminyear)**(1/n) - 1)*100, 2)
    data = {'Country': [country],
        'Min Year': [minyear],
        'Max Year': [maxyear],
        'GDP Min Year': [gdpminyear],
        'GDP Max Year': [gdpmaxyear],
        'Avg GDP': [avegdp],
        'CAGR (%)': [cagr]}
    
    return pd.DataFrame(data)

statdata(df1,'Afghanistan')
finaldf = pd.concat([statdata(df1,'Afghanistan'),statdata(df2,'Albania'),statdata(df3,'Algeria'),statdata(df4,'Angola')])
finaldf
```

```python
import pandas as pd

df_population = pd.DataFrame({
    "country": ["Finland", "Japan", "Bhutan", "Tuvalu"],
    "population": [5_620_000, 123_975_000, 790_000, 10_000]
})

df_gdp = pd.DataFrame({
    "country": ["Finland", "Japan", "Bhutan", "Tuvalu"],
    "gdp": [300, 4_200, 3.0, 0.06]
})

df_population 
df_gdp 

# Now run:pd.merge(df_population, df_gdp, on = "country")
pd.merge(df_population, df_gdp, on = "country")

# Write in simple terms what did you get? And how does this differ from pd.concat we used in other exercises?
'''
- We get a single DataFrame where the 2 columns 'country' are combined into a single column,
and 'gdp' column is added to the right side of the 'population' column.
- The 2 DataFrames are combined horizontally. 
- The indices are unique, in the order from 0 to 3.
- For merge: there must be a common 'country' column on both DataFrames!!
- Whereas, pd.concat will combine the 2 DataFrames vertically or horizontally depending on the variable 'axis = 0' or 'axis = 1'.
- Even if the pd.concat stacks the 2 tables horizontally (axis = 1), the 'country' will be added 2 times. 
Therefore, there will be 2 'country' variables/columns if we use 'pd.concat'.
- If the pd.concat stacks the 2 tables vertically (axis = 0), the country names will be added repeatedly in new rows.
The values for 'population' and 'gdp' will be either NaN or original numbers depending which original DataFrame they belonged to.
'''

pd.concat([df_population,df_gdp],axis=0)
```

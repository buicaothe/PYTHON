# Exercise 4 - Explore the Combined Data

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "country": ["Belize", "Liechtenstein", "Bhutan"],
    "population": [417_000, 40_000, 790_000]
})

df2 = pd.DataFrame({
    "country": ["Suriname", "São Tomé and Príncipe", "Tuvalu"],
    "population": [630_000, 240_000, 10_000]
})

df1
df2
```

```python
df = pd.concat([df1,df2],ignore_index=True)
df
```

```python
# Stack with pandas.concat and apply the following functions to _population_ column.
print('- Average/Mean value:',df.population.mean()) # The average population across all countries
print('- Sum value:',df.population.sum()) # The total population of all countries combined
print('- Minimum value:',df.population.min()) # The smallest population among the countries
print('- Maximum value:',df.population.max()) # The biggest population among the countries
print('- Count value:',df.population.count()) # The number of non-empty data rows/countries
print('- Median value:',df.population.median()) # The middle value when populations are sorted in order
print('- Standard Deviation value:',df.population.std()) # Standard deviation: shows how spread out the values are from the average
print('- Variance value:',df.population.var()) # Variance: measures the degree of spread in the data (squared standard deviation)
```

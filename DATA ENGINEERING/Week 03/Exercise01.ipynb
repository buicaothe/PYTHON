# COUNTRY LOOKUP

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("Afghanistan.csv")
df2 = pd.read_csv("Albania.csv")
df3 = pd.read_csv("Algeria.csv")
df4 = pd.read_csv("Angola.csv")
df5 = pd.read_csv("Armenia.csv")
```

```python
# Min Year, Max Year, GDG at Min Year, GDP at Max Year, Average GDP Growth over min year - max year, GDP average growth in percents
#1.  AFGHANISTAN:
minyear = df1["Year"].min()
maxyear = df1['Year'].max()
gdpminyear = round(df1.loc[df1["Year"].idxmin(),'GDP']/1e9, 2)
gdpmaxyear = round(df1.loc[df1["Year"].idxmax(),'GDP']/1e9, 2)
avegdp = round(df1['GDP'].mean()/1e9,2)
n = maxyear - minyear
cagr = round(((gdpmaxyear/gdpminyear)**(1/n) - 1)*100, 2)
print('Min Year:',minyear)
print('Max Year:',maxyear)
print('GDP at Min Year:',gdpminyear)
print('GDP at Max Year:',gdpmaxyear)
print('GDP average:',avegdp)
print('Average Growth %:',cagr)
```

```python
#2.  ALBANIA:
minyear = df2["Year"].min()
maxyear = df2['Year'].max()
gdpminyear = round(df2[df2["Year"] == minyear]['GDP'].iloc[0] / 1e9, 2)
gdpmaxyear = round(df2[df2["Year"] == maxyear]['GDP'].iloc[0] / 1e9, 2)
avegdp = round(df2['GDP'].mean()/1e9,2)
n = maxyear - minyear
cagr = round(((gdpmaxyear/gdpminyear)**(1/n) - 1)*100, 2)
print('Min Year:',minyear)
print('Max Year:',maxyear)
print('GDP at Min Year:',gdpminyear)
print('GDP at Max Year:',gdpmaxyear)
print('GDP average:',avegdp)
print('Average Growth %:',cagr)
```

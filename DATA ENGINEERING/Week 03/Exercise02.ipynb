# Exercise 2 - Learn Concat

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "country": ["Belize", "Liechtenstein", "Bhutan"],
    "population": [417_000, 40_000, 790_000]
})

df2 = pd.DataFrame({
    "country": ["Suriname", "São Tomé and Príncipe", "Tuvalu"],
    "population": [630_000, 240_000, 10_000]
})
print('- DataFrame 1 (df1):\n',df1)
print('- DataFrame 2 (df1):\n',df2)
df1
```

```python
# --------------------------------------------------------------
# A. Use pandas.concat to put them in one data frame called df.
# --------------------------------------------------------------
df = pd.concat([df1,df2])
df
```

```python
# --------------------------------------------------------------
# B. Use sort_values()-method to sort DataFrame df by population in ascending order.
# --------------------------------------------------------------
sort = df.sort_values('population')
sort
```

```python
# --------------------------------------------------------------
# C. What is the difference between pandas.concat([df1,df2], axis = 0) and pandas.concat([df1,df2], axis = 1)
# --------------------------------------------------------------
'''
- pandas.concat([df1,df2], axis = 0): APPEND ROWS OF DATAFRAMES (VERTICALLY)
- pandas.concat([df1,df2], axis = 1): APPEND COLUMNS OF DATAFRAMES (HORIZONTALLY)
'''

df_axis1 = pd.concat([df1,df2], axis = 1)
df_axis1
```

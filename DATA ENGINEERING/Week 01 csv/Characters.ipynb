```python
import pandas as pd
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('characters.csv')
# df = pd.read_csv("https://raw.githubusercontent.com/planetacomputer/starwars-full/main/characters.csv")
df.head()
```

```python
df.describe()
```

```python
print('- Histograms: Weight:')
plt.hist(df.weight, bins=50, color='g',edgecolor='blue')   
plt.xlabel('Weight (kg)', fontsize=16)
plt.ylabel("Freq")
plt.title("Matplotlib: Histogram of Weight", fontsize=18)
plt.show() 
```

```python
plt.hist(df.height, bins=50, color='r',edgecolor='blue')   
plt.xlabel('Height (m)', fontsize=16)
plt.ylabel("Freq")
plt.title("Matplotlib: Histogram of Height", fontsize=18)
plt.show() 
```

```python
sns.histplot(data = df, x= 'height', color='yellow', 
             edgecolor ='red',bins=10, hue = 'gender') 
plt.xlabel('Height', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of Characters' Height per Gender", fontsize=18)
plt.show()
```

```python
# make bar graph (QUALITATIVE)
sns.countplot(data = df, x='skin_color', hue = 'hair_color')
plt.title("Seaborn BAR CHART: ", fontsize=18)
plt.show()
```

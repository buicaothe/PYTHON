# Exercise 1 - Count the choices with value_counts

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("2026-DEAI-intro-data-engineering-people-preferences-dataset.csv",sep=";")
df.head()
```

### A) Run `value_counts()` on Favorite Food, Favorite Pet, and Music. For each column, what is the most common choice and how many people chose it?

```python
food = df.value_counts("Favorite Food")
print(food)
# just to check Type and DataType:
print("** Type of food variable:",type(food))
print("** Data type of DataFrame/Series:",food.dtype)
maxchoice = food.idxmax()
maxchoice = food.index[0]

maxnumber = food.max()
maxnumber = food.iloc[0]
maxnumber = food[maxchoice]
maxnumber = food.loc[maxchoice]
maxnumber = food.at[maxchoice]
maxnumber = np.max(food)

print(f'- Most common choice: {maxchoice} = {maxnumber} people')
```

```python
pet = df.value_counts("Favorite Pet")
print(pet)
# just to check Type and DataType:
print("** Type of pet variable:",type(pet))
print("** Data type of DataFrame/Series:",pet.dtype)
maxchoice = pet.idxmax()
maxchoice = pet.index[0] # due to sorted result

maxnumber = pet.max()
maxnumber = pet.iloc[0]
maxnumber = pet[maxchoice]
maxnumber = pet.loc[maxchoice]
maxnumber = pet.at[maxchoice]
maxnumber = np.max(pet)

print(f'- Most common choice: {maxchoice} = {maxnumber} people')
```

```python
music = df.value_counts("Music")
print(music)
# just to check Type and DataType:
print("** Type of music variable:",type(music))
print("** Data type of DataFrame/Series:",music.dtype)
maxchoice = music.idxmax()
maxchoice = music.index[0] # due to sorted result

maxnumber = music.max()
maxnumber = music.iloc[0]
maxnumber = music[maxchoice]
maxnumber = music.loc[maxchoice]
maxnumber = music.at[maxchoice]
maxnumber = np.max(music)

print(f'- Most common choice: {maxchoice} = {maxnumber} people')
```

### B) Run `value_counts(normalize=True)` on Favorite Season. What share of people chose summer?

```python
season = df.value_counts("Favorite Season",normalize=True)
print(season)

# option 1:
summer = round(season.Summer*100,2)
winter = round(season.Winter*100,2)
fall   = round(season.Fall*100,2)
spring = round(season.Spring*100,2)
# option 2:
summer = round(season["Summer"] * 100, 2)
winter = round(season["Winter"] * 100, 2)
fall = round(season["Fall"] * 100, 2)
spring = round(season["Spring"] * 100, 2)

print(f'- Share of Summer: {summer}%')
print(f'- Share of Winter: {winter}%')
print(f'- Share of Fall  : {fall}%')
print(f'- Share of Spring: {spring}%')

# option 3:
choice = []
for i in range(len(season)):
    value = round(season.iloc[i]*100,2)
    choice.append(value)
    print(f'- Share of {season.index[i]}: {choice[i]}%')


# just to check Type and DataType:
print("** Type of season variable:",type(season))
print("** Data type of DataFrame/Series:",season.dtype)
```

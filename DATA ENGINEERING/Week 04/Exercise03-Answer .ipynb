# Exercise 3 - Permutations

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("2026-DEAI-intro-data-engineering-people-preferences-dataset.csv",sep=";")
df.head()
```

```python
# Combination 1:
# "Your favorite color","Dream Holiday Destination","Favorite Pet"
combo1 = df[['Your favorite color','Dream Holiday Destination','Favorite Pet']]
count1 = combo1.value_counts()
filtercombo1 = count1[count1 > 1]
print('- COUNT FOR COMBO 1:\n',filtercombo1)

# Combination 2:
# "Your favorite color","Dream Holiday Destination","Favorite Season"
combo2 = df[['Your favorite color','Dream Holiday Destination','Favorite Season']]
count2 = combo2.value_counts()
filtercombo2 = count2[count2 > 1]
print('- COUNT FOR COMBO 2:\n',filtercombo2)

# Combination 3:
# "Your favorite color","Dream Holiday Destination","Music"
combo3 = df[['Your favorite color','Dream Holiday Destination','Music']]
count3 = combo3.value_counts()
filtercombo3 = count3[count3 > 1]
print('- COUNT FOR COMBO 3:\n',filtercombo3)

# Combine 3 tables with pandas.concat
combined_df = pd.concat([filtercombo1,filtercombo2,filtercombo3])

print("--- CONCAT RESULT ---")
print(combined_df)
```

```python
# --- BONUS: for all trios:
from itertools import permutations
columns = [
    "Your favorite color",
    "Dream Holiday Destination",
    "Favorite Pet",
    "Favorite Season",
    "Music",
    "Favorite Food",
]

trios = list(permutations(columns, 3))
results = []
for trio in trios:
    column_list = list(trio)
    
    counts = df[column_list].value_counts()
    filtered = counts[counts > 1].reset_index(name='count')
    if not filtered.empty:
        results.append(filtered)

if results:
    combined_bonus_df = pd.concat(results, ignore_index=True)
    print("--- COMBINATION OF ALL (COUNT > 1) ---")
    print(combined_bonus_df)
else:
    print("There is no count > 1!")
```

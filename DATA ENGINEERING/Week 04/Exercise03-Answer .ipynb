# Exercise 3 - Permutations

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("2026-DEAI-intro-data-engineering-people-preferences-dataset.csv",sep=";")
df.head()
```

```python
from itertools import permutations

# 1. Khởi tạo danh sách các bộ 3 cột
columns = [
    "Your favorite color",
    "Dream Holiday Destination",
    "Favorite Pet",
    "Favorite Season",
    "Music",
    "Favorite Food",
]

trios = list(permutations(columns, 3))
trios[:3]

# --- PHẦN 1: Làm thủ công cho 3 bộ permutations đầu tiên ---
selected_trios = trios[:3]
results = []

for trio in selected_trios:
    # df[list(trio)]: trích xuất 3 cột tương ứng
    # value_counts(): đếm số lần kết hợp giữa các giá trị trong 3 cột
    counts = df[list(trio)].value_counts()
    
    # Lọc các kết hợp xuất hiện hơn 1 lần
    filtered = counts[counts > 1]
    results.append(filtered)

# Gộp kết quả lại bằng pandas.concat
combined_df = pd.concat(results)
print(combined_df)

# --- BONUS: Viết vòng lặp duyệt qua TẤT CẢ các bộ trong trios ---
all_results = []
for trio in trios:
    counts = df[list(trio)].value_counts()
    filtered = counts[counts > 1]
    
    # Chỉ thêm vào danh sách nếu có kết hợp > 1
    if not filtered.empty:
        all_results.append(filtered)

final_bonus = pd.concat(all_results)
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("2026-DEAI-intro-data-engineering-people-preferences-dataset.csv",sep=";")
df.head()
```

### A) Keep only the rows where Favorite Pet is dog. Run `value_counts(normalize=True)` on Dream Holiday Destination. Which category changes most which least?  

```python
dog = df[df['Favorite Pet']=='Dog']
dog.head(3)
# just to check Type and DataType:
print("** Type of food variable:",type(dog))
print("** Data type of DataFrame:\n",dog.dtypes)
print("** DataFrame info:\n",dog.info())
```

```python
holiday = dog.value_counts('Dream Holiday Destination', normalize=True)
print(holiday)

choice = []
for i in range(len(holiday)):
    value = round(holiday.iloc[i]*100,2)
    choice.append(value)
    print(f'- Share of {holiday.index[i]}: {choice[i]}%')

maxchoice = holiday.idxmax()
maxnumber = round(holiday[maxchoice]*100,2)
print(f'- Most common choice: {maxchoice} = {maxnumber}%')

# just to check Type and DataType:
print("** Type of music variable:",type(holiday))
print("** Data type of DataFrame/Series:",holiday.dtype)
```

### B) Do the same analysis for Favorite Pet equal to cat. Which destination shows up for cats and did not show up for dogs?

```python
print('- Holiday for Cat:\n',holidaycat)
print('- Holiday for Dog:\n',holiday)

set_cat = set()
for i in range(len(holidaycat)):
    set_cat.add(holidaycat.index[i])

set_dog = set()
for i in range(len(holiday)):
    set_dog.add(holiday.index[i])

catdes = list(set_cat - set_dog)
print('- Destinations show up only in Cat:', catdes)
```

### C) Keep only the rows where Favorite Season is summer, and run `value_counts(normalize=True)` on Favorite Pet. Then do the same for winter. In both seasons, which pet is the most common?

```python
# Favorite Season = Summer:
summer = df.loc[df['Favorite Season'] == 'Summer']
summer = df[df['Favorite Season'] == 'Summer']
summer.head()
```

```python
winter = df.loc[df['Favorite Season']=='Winter']
winter = df[df['Favorite Season']=='Winter']
winter.head()
```

```python
petsummer = summer.value_counts('Favorite Pet',normalize=True)
petsummer.name = 'Summer'
print('A) Favorite pet for favorite summer:\n',petsummer)
petwinter = winter.value_counts('Favorite Pet',normalize=True)
petwinter.name = 'Winter'
print('B) Favorite pet for favorite winter:\n',petwinter)

# just to check Type and DataType:
print("** Type of food variable:",type(petsummer))
print("** Data type of DataFrame/Series:",petsummer.dtype)
```

```python
# Which pet is the most common?
## FAVORITE SUMMER:
def checkcommon(petseason):
    maxpercent = round(petseason.max()*100,2)
    mostcommon = petseason.index[petseason == maxpercent]
    mostcommon = petseason.idxmax()
    print(f'- Most common pet for favorite {petseason.name}: {mostcommon} = {maxpercent}%')

checkcommon(petsummer)

## FAVORITE WINTER:
checkcommon(petwinter)

## COMPARING SUMMBER & WINTER:
def common_for_both():
    for i in range(len(petsummer)):
        for j in range(len(petwinter)):
            if petsummer.index[i] == petwinter.index[j]:
                return print('--> Most common pet for both Summer and Winter: ',petsummer.index[i])
    return print('--> There is no common pet between Summer and Winter!')

common_for_both()
```

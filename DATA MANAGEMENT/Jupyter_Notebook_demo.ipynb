```python
# This is how you import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as plt
```

```python
# Load dataset, let's use the Iris dataset: https://www.kaggle.com/datasets/uciml/iris.
# Data is loaded to variable iris.
iris = sns.load_dataset("iris")
```

```python
# Show first few rows of the dataset
# head() shows 5 first lines, but head(10) shows 10 first lines

iris.head(10)
```

```python
# tail() shows five last lines of the dataset
iris.tail()
```

```python
# This command shows the number of rows and columns. You get the same info by using the info(), see below.
iris.shape
```

```python
# This is a very usefull command. Learn it!
iris.info()
```

```python
# Mean, mode, median etc. can be calculated for each variable (column)...
np.mean(iris["sepal_length"])
```

```python
# ... but learn also this describe()-function, because it does all the measures for every column.
iris.describe()
```

## Task 2: Explain the code (add explanation as comments to the code cells) (1 point)
- Do not change the code, only add comments.

```python
# Explain this code

cols = df.columns # show the all columns names, assigning this to variable 'cols'
print(cols)
```

```python
# Explain this code

for col in cols:
    df[col] = df[col].apply(lambda x: x.replace(',', '.') if isinstance(x, str) else x)
    # Explain: for each names in cols (apply() is applied to each element x in col):
    # If the value is string (isinstance (x,str)), change ',' (if any) into '.'
    # If not a string, then keep original x. This is to ensure not to take the numeric or NaN values, resulting in errors.
    # at this step, the string is not yet numeric.
    # This step is just for standardize the format, ready for converting from string to numeric in next steps.

df.head(10) # show first 10 rows
```

### Let's do some conversion here.

```python
# Convert data types to numeric in selected columns.
# Do not change this code.
cols = ['Distance', 'Calories', 'Avg HR', 'Max HR', 'Aerobic TE', 'Avg Cadence', 'Max Cadence',
       'Avg Pace', 'Best Pace', 'Total Ascent', 'Total Descent',
       'Avg Stride Length', 'Avg Cadence.1', 'Max Cadence.1',
       'Training Stress Score®', 'Total Strokes', 'Avg. Swolf',
       'Avg Stroke Rate', 'Steps', 'Min Temp', 'Decompression',
       'Number of Laps', 'Max Temp', 'Min Elevation', 'Max Elevation'] # 
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')
# apply to each value in cols and change to Numbers.
# errors='coerce': if there is no number (special characters, letters, or empty string, or -), change to NaN value
df.info()
```

## Plotting quantitative data
- Histogram is used to show the distribution of quantitative data
- plot.hist using Matplotlib (https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html)
- sns.histplot using Seaborn (https://seaborn.pydata.org/generated/seaborn.histplot.html)

```python
plt.hist(trips.Age, bins=10, color='g',edgecolor='blue')   
plt.xlabel('Age', fontsize=16)
plt.ylabel("Freq")
plt.title("Matplotlib: Histogram of travellers' age", fontsize=18)
plt.show() 
```

### Histogram with Seaborn
- Seaborn is another library for graphics
- Based on Matplotlib
- histplot is a function in Seaborn for Historgrams 
- https://seaborn.pydata.org/generated/seaborn.histplot.html

```python
sns.histplot(data = trips, x= 'Age', color='yellow', 
             edgecolor ='red',bins=10, hue = 'Gender') 
plt.xlabel('Age', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of travellers' age per Gender", fontsize=18)
plt.show()
```

##### Note: Using parameter multiple you get separate bars per variable you have in hue parameter

```python
sns.histplot(data = trips, x= 'Age', color='yellow', 
             edgecolor ='red',bins=10, hue = 'Gender', multiple = 'dodge') 
plt.xlabel('Age', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of travellers' age per Gender", fontsize=18)
plt.show()
```

Modify the histogram's parameters and labels in order to see, what do they do.

```python
# Modify some parameters

sns.histplot(data = trips, x= 'Age', color='yellow', 
             edgecolor ='red',bins=10, hue = 'Gender', multiple = 'dodge') 
plt.xlabel('Age', fontsize=14)
plt.ylabel('Freq', fontsize=12)   
plt.title("Seaborn: Histogram of travellers' age per Gender", fontsize=18)
plt.show()
```

### Task 5
- What is the difference between a histogram and bar chart?

#### 1. HISTOGRAM:
- Used to show the distribution of "quantitative" data
- Data type: Continuous numerical data (weight, height, temperature...)
- X-axis: A continuous numeric scale divided into intervals or bins
- Bar scapcing: Bars touch each other to indicate a continuous range
- Order of bar: Must follow strict numerical order
#### 2. BAR CHART:
- Used to show the distribution of "qualitative" data
- Data type: Discrete or categorical data (countries, car brands, departments...)
- X-axis: Distinct, separate categories or names
- Bar scapcing: Bars have spaces between them to show independent categories
- Order of bar: Can be rearranged (by frequency, size, or alphabetically...) without losing meaning

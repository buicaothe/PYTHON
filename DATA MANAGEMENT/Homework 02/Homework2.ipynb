```python
# 2. Create a crosstab sex x survived (change the column indexes) and normalize over each row.
sex_survive = pd.crosstab(df['sex'],df['survived'],
            rownames =['Sex'], colnames = ['Survival'],
            normalize='index', # normalize for each row
            margins=True, margins_name="Total") # row grouping = sex; columns = Survival; aggregation sum() default.
sex_survive.columns = sex_survive.columns.map({0:'No',1:'Yes'}) # change columns names from '0' to 'No' and from '1' to 'Yes'
round(sex_survive,2)
```

#### Task 5.6: Explain why the mean and std are what they are after scaling.

**Mean = 0 and Std = 1**
----
##### 1. Why Mean = 0:

* The formula used by StandardScaler to transform each value x is:


$$z = \frac{x - \mu}{\sigma}$$


* Subtracting the mean slides the entire distribution along the number line so its center rests squarely at 0.


##### 2. Why Standard Deviation = 1:

* Dividing by the standard deviation shrinks or expands the scale so the overall spread measures exactly 1.

## Task 6: Linear regression (1 point each = 5 points in total)

1. Use age as the independent variable X and fare as the dependent variable y and fit a linear regression model.
2. Print out the slope and intercept, use 2 decimal precision.
3. Plot a scatter plot with a regression line.
4. Print out the R^2 (R squared) value. Hint: score(X,y). Use 5 decimal precision.
5. Based on the scatterplot and the R^2 value, do you think there is a linear relationship between the age and fare? Why/why not? Also explain, what is the R^2 value.

```python
# Take Linear regression in use from Scikit-learn
from sklearn.linear_model import LinearRegression
# import scipy.stats as stats 
# library for statistical measurementssns.get_dataset_names()
```

```python
# create vectors containing the data
# x independent variable
x = df5['age'].values # this is numpy.ndarray (needs to convert to VECTOR)
# df5['age'] #  this is a pandas Series
# print(type(df5['age'].values)) 
# print(type(df5['age']))
x
```

```python
x.shape # x is a normal array (numpy)
```

```python
# Create vector x containing the data 'age' as 1-dimensional ndarray
x = df5['age'].values.reshape(-1,1)
print("shape:", x.shape)
# show 10 first items of the array x
print("first 10 values of x:\n", x[0:10])
```

```python
# Show the last 10 values of the array x using slicing
print("last 10 values of x:", x[-10:])
```

```python
# create vector y containing the data 'tip'
y = df5['fare'].values
# show 20 first items of the array y
y [0:20]
```

```python
# Create linear regression model
model = LinearRegression()
# fit the data
model.fit(x, y)
# calculate the coefficients
slope = model.coef_[0]
intercept = model.intercept_
print(f"Slope: {slope}")
print(f"Intercept: {intercept}")
```

```python
# Visualise the results
# show the scatter diagram for the data and the linear regression line
plt.scatter(x, y, label='Data')
plt.plot(x, model.predict(x), color='red', label='Regression line')
plt.xlabel('x = Age')
plt.ylabel('y = Fare')
plt.title('Relationship between Age and Fare')
plt.legend()
plt.show()
```

```python
# 4. Print out the R^2 value
r2_score = model.score(x, y)
print(f"R^2 score: {r2_score:.5f}")
```

Task 6.5: Analysis and Interpretation:
- Is there a linear relationship?
--> No, there is no meaningful linear relationship between age and fare.
- Why? (Based on scatterplot and $R^2$):
    * Scatterplot: The data points are heavily scattered without forming any clear upward or downward linear trend. Most fares are clustered at the bottom regardless of age, with several extreme outliers (very expensive tickets) spread across different age groups.
    * $R^2$ value: The $R^2$ score is extremely close to 0 (around 0.01 or ~1%). This indicates that less than 1% of the variability in fare can be explained by age.

- What is the $R^2$ value?
    * $R^2$ (the coefficient of determination) measures the proportion of variance in the dependent variable (y, fare) that is predictable or explained by the independent variable (x, age). A value near 1 indicates a strong linear fit, while a value near 0 indicates the linear model fails to explain the variance in the target variable.

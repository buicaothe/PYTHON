## Task 3: CrossTab (1 point each = 2 points in total)
1. Create a crosstab sex x survived, use .map({0: "No", 1: "Yes"}) to change the column indexes.
2. Create a crosstab sex x survived (change the column indexes) and normalize over each row. Round to 2 decimal precision. Hint: Check the correct parameter and its value from here https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html

```python
# 1. Create a crosstab sex x survived, use .map({0: "No", 1: "Yes"}) to change the column indexes.
sex_survive = pd.crosstab(df['sex'],df['survived'],
            rownames =['Sex'], colnames = ['Survival'],
            margins=True, margins_name="Total") # row grouping = sex; columns = Survival; aggregation sum() default.
sex_survive.columns = sex_survive.columns.map({0:'No',1:'Yes'}) # change columns names from '0' to 'No' and from '1' to 'Yes'
sex_survive
```

```python
# 2. Create a crosstab sex x survived (change the column indexes) and normalize over each row.
sex_survive = pd.crosstab(df['sex'],df['survived'],
            rownames =['Sex'], colnames = ['Survival'],
            normalize='index', # normalize for each row
            margins=True, margins_name="Total") # row grouping = sex; columns = Survival; aggregation sum() default.
sex_survive.columns = sex_survive.columns.map({0:'No',1:'Yes'}) # change columns names from '0' to 'No' and from '1' to 'Yes'
round(sex_survive,2)
```

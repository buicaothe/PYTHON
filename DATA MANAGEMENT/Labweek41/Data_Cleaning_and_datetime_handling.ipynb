Removing characters with replace

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('[^a-zA-Z0-9]','', regex=True)
# REPLACE [^a-zA-Z0-9]: all characters which is NOT alphabet characters (from a-z, or A-Z) or numbers (from 0-9), with '' : which means delete all special charaters
# Regex = True: Regular Expression the pattern should be interpreted as a regular expression (regex) rather than as ordinary text
```

```python
df
```

Changing format of string data

```python
# Data converted first to string
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: str(x)) # use lambda function to convert number to string
# Then new format for data
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
```

```python
df
```

```python
# Clean Phone Number
df["Phone_Number"] = df["Phone_Number"].str.replace('Na--','')
df
```

## Remove other strange values from Phone_Number (nan--, ...)

```python
df['Phone_Number']=df['Phone_Number'].str.replace('nan--','',regex=True)
df['Phone_Number']=df['Phone_Number'].str.replace('nan--','')
df
```

## Split column data in several new variables
- split Address into "Street_Address", "State" and "Zip_Code" columns.
- use str.split

```python
df[['Street_Address','State','Zip_Code']]=df['Address'].str.split(',',n=2,expand=True)
# Split column "Address" into 3 new columns "Street_Address", "State" and "Zip_Code".
# Separation is ','
# n=2: the number of 'cut'. So if n = 2 (the first 2 ','), there would be 2 cuts, which results in 3 columns
# expand = True: create new columns
# However, the original 'Address' column is still there in the df.
df # print and check the new columns
```

## Fill all NaN (Not a Number) values with ''
- '' means an empty value, i.e. in this case fillna means just removing all NaNs.

```python
df=df.fillna('') # Notes: 'NaN' is not a normal text/string in the dateframe.
# df1['Last_Name']=df1['Last_Name'].str.replace('NaN','')
df
```

## Continue cleaning and change values for 'Do_Not_Contact' and 'Paying Customer' to Yes = Y, No = N

```python
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes','Y')
df['Paying Customer'] = df['Paying Customer'].str.replace('No','Y')
df
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes','Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No','Y')
df
```

## Current Time

```python
# Take ctime in use
from time import ctime
cur_time = time()
print("Current time:", ctime(cur_time))
print("Universal time:",time())
```

```python
# Check some past times
past_time = 0
print("Starting time from", ctime(past_time), "as the position", past_time)

past_time = 1
print("Time in position",  past_time, "is:", ctime(past_time))

past_time = 12345
print("Time in position",  past_time, "is:", ctime(past_time))
```

## Access attibutes in datetime

```python
# Load current time
today = datetime.now()
print("time now is", today)
```

```python
# Print year
print("current year:", today.year)

# Print month
print("current month:", today.month)

# Print day and hour
print("current day:", today.day, "and hour:", today.hour)
```

## Converting strings to Datetime object

```python
# Define date as string
date_str = "2023-11-20 12:30:15"

# Format
format_str = "%Y-%m-%d %H:%M:%S"
date2 = datetime.strptime(date_str, format_str)
print("original date:", date_str, "in format", type(date_str))
print("formatted date:", date2, "in format", type(date2))
```

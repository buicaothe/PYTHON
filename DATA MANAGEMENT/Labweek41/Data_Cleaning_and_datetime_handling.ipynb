# Time stamps and Datetime handling

## Timestamp

```python
# Take libraries for time management in use
from time import time, time_ns # time: total secenods from beginning
print("time:", time()) # time: total secenods from beginning
print("time in ns:", time_ns()) # time_ns: total time in nano-seconds
```

## Current Time

```python
# Take ctime in use
from time import ctime # ctime: Date-month-time-year
cur_time = time() # use ctime(time()) to take the current time
print("Current time:", ctime(cur_time))
print("Universal time:",time())
```

```python
# Check some past times
past_time = 0
print("Starting time from", ctime(past_time), "as the position", past_time) # at exactly 2:00:00 on 01/01/1970

past_time = 1
print("Time in position",  past_time, "is:", ctime(past_time)) # 1-second difference: at exactly 2:00:01 on 01/01/1970

past_time = 12345
print("Time in position",  past_time, "is:", ctime(past_time)) # 12345-second difference: at exactly 5:25:45 on 01/01/1970 (about 5.5 hours)
```

### 1. READ EXCEL SHEETS:

```python
# read from Excel file "Steps.xlsx":
steps1 = pd.read_excel('Steps.xlsx',sheet_name='Steps1') # read sheet name 'Steps1' the excel file
steps2 = pd.read_excel('Steps.xlsx',sheet_name='Steps2') # read sheet name 'Steps2' the excel file
steps3 = pd.read_excel('Steps.xlsx',sheet_name='Steps3') # read sheet name 'Steps3' the excel file
```

```python
steps1.head(2) # print and see the first 2 rows
```

```python
steps2.head(2) # print and see the first 2 rows
```

```python
steps3.head(2) # print and see the first 2 rows
```

### 2. CONCAT THE 3 DATAFRAMES:

```python
# Combine the data in single DataFrame (using Pandas concat)
allsteps = pd.concat([steps1,steps2,steps3]) # combine vertically the 3 DataFrames in to single DataFrame without resetting the Indices.
allsteps = pd.concat([steps1,steps2,steps3],ignore_index=True) # combine vertically the 3 DataFrames in to single DataFrame with resetting the Indices (using ignore_index).
allsteps # print to see the overall of the newly concatted DataFrame
```

### 3. New columns in DataFrame for 'year', 'month' and 'day' update the DataFrame columns to contain: 'year', 'month', 'day' and 'steps'

```python
allsteps['year'] = allsteps['Date'].dt.year # split the 'year' from the "Date" column with accessor .dt.year
allsteps['month'] = allsteps['Date'].dt.month # split the 'month' from the "Date" column with accessor .dt.month
allsteps['day'] = allsteps['Date'].dt.day # split the 'day' from the "Date" column with accessor .dt.day
allsteps
```

### 4. Make boxplot where there is for each month amount of steps separated by year

```python
sns.boxplot(data=allsteps, x='month', y='Steps', hue='year',palette='Set2') 
# print the boxplot using seaborn library: x: month; y for total steps. HUE = year, use color pallets Set2
plt.xlabel('Month of the Year',fontsize=10) # give the name of the x axis
plt.ylabel('Steps',fontsize=10) # give the name of the y axis
plt.title('Amount of steps Each Month separated by year', fontsize=12) # give the title of the Graph
plt.show # print the graph
```

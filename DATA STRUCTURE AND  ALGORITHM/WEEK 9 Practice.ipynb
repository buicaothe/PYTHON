# Task 1 - Implementing a Python stack



A stack is a linear data type in which new items are added to/ deleted from top (LIFO). The functions associated with stack are:

|function    | Description                                             |Time complexity|
|------------|---------------------------------------------------------|---------------|
|**empty()** | Returns whether the stack is empty                      | O(1)          |
|**size()**  | Returns the size of the stack                           | O(1)          |
|**top()**   | Returns a reference to the topmost element of the stack | O(1)          |
|**push(a)** | Inserts the element ‘a’ at the top of the stack         | O(1)          |
|**pop()**   | Deletes the topmost element of the stack                | O(1)          |

Implement and demonstrate the use of your Python stack data structure e.g. by using Python **list**, **Collections.deque** or **queue.LifoQueue**. Analyse critically the implementation (pros and cons).


```python
class Stack: # Object-Oriented Programming
    def __init__(self):
        self.items = [] # initiate an empty list
    def is_empty(self):
        return len(self.items)==0 # to check if the stack is empty or not by comparing length of list with 0 
    def push(self,item):
        self.items.append(item) # push the element/item: adding to right side (last position), which always take -1 index
    def pop(self):
        if not self.is_empty(): # to make sure the list is not empty before popping, otherwise resulting in error.
            return self.items.pop() # by default removing the last element (-1) LIFO
        print('Stack is empty')

    def peek(self):
        if not self.is_empty(): # to make sure the list is not empty before popping, otherwise resulting in error.
            return self.items[-1] # take the last item
        print('Stack is empty') 

    def size(self):
        return len(self.items) # size = length of the list

stack = Stack()
print("Is the stack empty?",stack.is_empty())
stack.push(200)
stack.push(400)
stack.push(600)
print('- Stack size:',stack.size())
print('- Stack top element - peek:',stack.peek())
print('- Pop item from stack:',stack.pop())
print('- Stack size after pop:',stack.size())
```

A stack is a linear data type in which items are stored in FIFO manner. The functions associated with stack are:

|function    | Description                                                                                          |Time complexity|
|------------|------------------------------------------------------------------------------------------------------|---------------|
|**rear**    | Get the last item from queue                                                                         | O(1)          |
|**front**   | Get the first item from queue                                                                        | O(1)          |
|**enqueue** | Adds an item to the queue. If the queue is full, then it is said to be an Overflow condition         | O(1)          |
|**dequeue** | Removes an item from the queue. The items are popped in the same order in which they are pushed.     |               |
|            | If the queue is empty, then it is said to be an Underflow condition                                  | O(1)          |

Implement and demonstrate the use of your Python stack data structure e.g. by using Python **list**, **Collections.deque** or **queue.LifoQueue**. Analyse critically the implementation (pros and cons).


```python
class Queue: # Object-Oriented Programming
    def __init__(self):
        self.items = []
    def is_empty(self):
        return len(self.items)==0
    def enqueue(self,item):
        self.items.append(item)
    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0) # remove the first element with index-0 (FIFO)
        else:
            print('Queue is empty')
            return None

    def peek(self):
        if not self.is_empty():
            return self.items[0]
        else:
            print('Queue is empty')
            return None
    def size(self):
        return len(self.items)

queue = Queue()
print("Is the queue empty?",queue.is_empty())
queue.enqueue(200)
queue.enqueue(400)
queue.enqueue(600)
print('- Queue size:',queue.size())
print('- Queue top element - peek:',queue.peek())
print('- Pop item from queue:',queue.dequeue())
print('- Queue size after pop:',queue.size())
print(queue.items)
```

### 3. USER-DEFINED DATA STRUCTURE FOR A SINGLE APPLICANT:
For a single applicant, a class called 'Applicant' is used: class Applicant includes the following needed Attributes:
- Name
- grades: Math, Finnish, English
- school_grade_average

METHODS (FUNTIONALITTIES):
- calculate_weighted_average: this is calculate the weighted average based on Math, Finnish and English
- update_grade: update the school grades of an applicant
- update_school_grade_average
- get_details : formatted string with all the details of applicant

```python
# OBJECT-ORIENTED PROGRAMMING:
class Applicant:
    def __init__(self, name, age, residence, siblings, height, hobby, grades, school_grade_average):
        # 1. personal information attributes:
        self.name = name
        self.age = age
        self.residence = residence
        self.siblings = siblings
        self.height = height
        self.hobby = hobby
        
        # 2. Attributes for application process (functionalities):
        self.grades = grades  # dict: {'Math': 8, 'Finnish': 9, 'English': 6}
        self.school_grade_average = school_grade_average

    def calculate_weighted_average(self):
        # Finnish x 2, Math x 1.5, English x 1.5: 
        math_score = self.grades.get("Math", 0)
        finnish_score = self.grades.get("Finnish", 0)
        english_score = self.grades.get("English", 0)
        
        total_points = (math_score * 1.5) + (finnish_score * 2.0) + (english_score * 1.5)
        total_weights = 1.5 + 2.0 + 1.5
        return total_points / total_weights

    def update_grade(self, subject, grade):
        # update grade for a specific subject:
        if subject in self.grades:
            self.grades[subject] = grade

    def update_school_grade_average(self, new_average):
        # update school_grade_average:
        self.school_grade_average = new_average

    def get_details(self):
        weighted_avg = self.calculate_weighted_average()
        return (f"Applicant: {self.name} | School Avg: {self.school_grade_average} "
                f"| Weighted Avg: {weighted_avg:.2f} | Grades: {self.grades}")

    def __lt__(self, other):
        return self.calculate_weighted_average() < other.calculate_weighted_average()
```

### 4. COLLECTION OF APPLICANTS DATA STRUCTURE AND FUNCTIONALITIES:
For storing a collection of Application objects, a suitable data structure is a List or a Dictionary.

Functionalities:
- Add Applicant
- Remove Applicant Details
- Caculate Average
- Find Best Applications
- Display All Applicants
- Search Applicant
- Sort Applicants: search applicants based on a certain attribute (age, grade...)

###  A. Collection of applicants & Functionalities:

```python
class ApplicantCollection:
    def __init__(self):
        self.applicants = {}

    def add_applicant(self, applicant):
        """1. ad new Applicant to list"""
        self.applicants[applicant.name] = applicant

    def remove_applicant(self, name):
        """2. Delete an Applicant from list"""
        if name in self.applicants:
            del self.applicants[name]
        else:
            print(f"Applicant {name} not found.")

    def update_applicant_details(self, name, subject=None, new_grade=None, new_school_avg=None):
        """3. update grades"""
        applicant = self.search_applicant(name)
        if applicant:
            if subject and new_grade is not None:
                applicant.update_grade(subject, new_grade)
            if new_school_avg is not None:
                applicant.update_school_grade_average(new_school_avg)
        else:
            print(f"Applicant {name} not found.")

    def calculate_average(self):
        """4. for all applicants"""
        if not self.applicants:
            return 0.0
        total_school_avg = sum(app.school_grade_average for app in self.applicants.values())
        return total_school_avg / len(self.applicants)

    def find_best_applicant(self):
        if not self.applicants:
            return None
        return max(self.applicants.values(), key=lambda app: app.calculate_weighted_average())

    def display_all_applicants(self):
        if not self.applicants:
            print("No applicants in the collection.")
            return
        for app in self.applicants.values():
            print(app.get_details())

    def search_applicant(self, name):
        """7. Search based on name"""
        return self.applicants.get(name, None)

    def sort_applicants(self, attribute="weighted_average", reverse=True):
        """Sorting"""
        applicant_list = list(self.applicants.values())
        if attribute == "weighted_average":
            applicant_list.sort(key=lambda app: app.calculate_weighted_average(), reverse=reverse)
        elif attribute == "school_grade_average":
            applicant_list.sort(key=lambda app: app.school_grade_average, reverse=reverse)
        elif attribute == "age":
            applicant_list.sort(key=lambda app: app.age, reverse=reverse)
        elif attribute == "name":
            applicant_list.sort(key=lambda app: app.name, reverse=reverse)
        return applicant_list
```

For example, you can create a program to **manage a library's collection of books**. In this example, you could create classes for books, authors, publishers, and library staff. Create a custom linked list class to represent the books in the library's collection. The program should be able to perform basic operations such as adding books to the collection, searching for books by author or title, and checking books in and out. 

### CLASS DIAGRAM - Library Management System

```python
from abc import ABC, abstractmethod
from typing import Optional


class Item(ABC):
    def __init__(self, item_id: str, title: str):
        self._id = str(item_id)
        self._title = str(title)
        self._is_borrow = False

    @abstractmethod
    def get_details(self) -> str:
        pass


class Book(Item):
    def __init__(self, isbn: str, title: str, author: str):
        super().__init__(item_id=isbn, title=title)
        self._isbn = str(isbn)
        self._author = str(author)

    def get_details(self) -> str:
        status = "Borrow" if self._is_borrow else "Available"
        return f"[ISBN: {self._isbn}] '{self._title}' - Author: {self._author} | Status: {status}"


class BookNode:
    def __init__(self, book: Book):
        self.book = book
        self.next: Optional['BookNode'] = None


class BookLinkedList:
    def __init__(self):
        self._head: Optional[BookNode] = None
        self._count: int = 0

    def append(self, book: Book) -> None:
        new_node = BookNode(book)
        if self._head is None:
            self._head = new_node
        else:
            current = self._head
            while current.next is not None:
                current = current.next
            current.next = new_node
        self._count += 1

    def find_by_isbn(self, isbn: str) -> Optional[Book]:
        current = self._head
        while current is not None:
            if current.book._isbn == str(isbn):
                return current.book
            current = current.next
        return None

    def display_all(self) -> None:
        if self._head is None:
            print("Empty book list")
            return
        current = self._head
        while current is not None:
            print("  *", current.book.get_details())
            current = current.next


class LibraryManager:
    def __init__(self):
        self._catalog = BookLinkedList()

    def check_out(self, isbn: str) -> bool:
        book = self._catalog.find_by_isbn(isbn)
        if book is not None and not book._is_borrow:
            book._is_borrow = True
            return True
        return False

    def return_item(self, isbn: str) -> bool:
        book = self._catalog.find_by_isbn(isbn)
        if book is not None and book._is_borrow:
            book._is_borrow = False
            return True
        return False


class ConsoleUI:
    def __init__(self, manager: LibraryManager):
        self.manager = manager

    def render_menu(self) -> None:
        print("\n--- LIBRARY MENU ---")
        print("1. See all books")
        print("2. Borrow books")
        print("3. Return books")
        print("0. Exit")

    def get_user_input(self) -> str:
        return input("Your slection: ").strip()
```
